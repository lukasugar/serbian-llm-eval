```python
def get_indexes(l, key, value):
    return [i for i, v in enumerate(l) if v[key] == value]
```

```python
def check_if_lists_are_equal(list1, list2):
    """Check if each element in the lists are equal"""
    for i in range(len(list1)):
        if list1[i] != list2[i]:
            return False
    return True

```

```python
a = [{"a": 1}, {"a": 2}, {"a": 3}, {"a": 1}]
```
